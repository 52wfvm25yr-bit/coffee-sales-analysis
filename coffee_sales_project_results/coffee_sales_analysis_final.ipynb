## Описание данных

В проекте используется датасет Coffee Sales с платформы Kaggle.

Исходный файл `Coffe_sales.csv` содержит 3 547 строк и 11 столбцов. Данные охватывают период с 1 марта 2024 года по 23 марта 2025 года.

В таблице представлены название напитка, показатель `money`, дата и время записи, час, часть дня, день недели и месяц.

Одна строка рассматривается как одна запись в датасете. В данных нет отдельного идентификатора заказа, поэтому количество строк нельзя уверенно называть количеством уникальных транзакций.

Точный экономический смысл и валюта столбца `money` не подтверждены.

```python
import pandas as pd

# Выбираем числовые столбцы
numeric_columns = df.select_dtypes(include="number").columns.tolist()

# 1. Основная статистика по числовым столбцам
print("Описательная статистика:")
display(df[numeric_columns].describe().T)

# 2. Минимальные и максимальные значения
min_max_summary = pd.DataFrame({
    "min": df[numeric_columns].min(),
    "max": df[numeric_columns].max()
})

print("Минимумы и максимумы:")
display(min_max_summary)

# 3. Ищем отрицательные значения в числовых столбцах
negative_summary = pd.DataFrame({
    "negative_count": [
        (df[column] < 0).sum()
        for column in numeric_columns
    ]
}, index=numeric_columns)

print("Количество отрицательных значений:")
display(negative_summary)

# 4. Проверяем допустимые диапазоны для известных полей
invalid_ranges = pd.DataFrame({
    "Проверка": [
        "money <= 0",
        "hour_of_day вне диапазона 0–23",
        "Weekdaysort вне диапазона 1–7",
        "Monthsort вне диапазона 1–12"
    ],
    "Количество строк": [
        (df["money"] <= 0).sum(),
        (~df["hour_of_day"].between(0, 23)).sum(),
        (~df["Weekdaysort"].between(1, 7)).sum(),
        (~df["Monthsort"].between(1, 12)).sum()
    ]
})

print("Проверка допустимых диапазонов:")
display(invalid_ranges)

# 5. Выбираем текстовые и категориальные столбцы
category_columns = df.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

# Показываем значения только для столбцов
# с небольшим количеством категорий
print("Уникальные значения категориальных столбцов:")

for column in category_columns:
    unique_count = df[column].nunique(dropna=False)

    if unique_count <= 20:
        print(f"\n{column}: {unique_count} уникальных значений")
        print(df[column].value_counts(dropna=False))

# 6. Ищем статистические выбросы методом IQR
outlier_summary = []

for column in numeric_columns:
    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        "column": column,
        "lower_bound": round(lower_bound, 2),
        "upper_bound": round(upper_bound, 2),
        "outlier_count": outlier_count
    })

outlier_summary = pd.DataFrame(outlier_summary)

print("\nПотенциальные выбросы по методу IQR:")
display(outlier_summary)
```

## Подготовка и очистка данных

Перед анализом создаётся отдельная рабочая копия `df_clean`. Исходный DataFrame сохраняется без изменений для контрольной проверки.

В ходе подготовки:

- столбец `Date` преобразуется в формат даты;
- дата и время объединяются в один временной признак;
- создаётся календарный период `year_month`;
- час дополнительно извлекается из значения `Time`;
- проверяется согласованность `hour_of_day` и исходного времени;
- создаётся признак выходного дня;
- проверяется полнота календарных месяцев.

Строки намеренно не удаляются, пропущенные значения не заполняются, исходный столбец `money` не изменяется.

Производные признаки создаются только для проверки качества и дальнейших группировок.


```python
import pandas as pd

# Сохраняем копию исходного DataFrame для контрольной проверки
df_original_snapshot = df.copy(deep=True)

# Создаём рабочую копию для подготовки данных
df_clean = df.copy()

# Преобразуем дату в отдельный столбец, сохраняя исходный Date
df_clean["date_parsed"] = pd.to_datetime(
    df_clean["Date"],
    format="%Y-%m-%d",
    errors="coerce"
)

# Объединяем исходные дату и время в одну временную метку
# format="mixed" позволяет обработать время с микросекундами и без них
df_clean["transaction_datetime"] = pd.to_datetime(
    df_clean["Date"].astype(str)
    + " "
    + df_clean["Time"].astype(str),
    format="mixed",
    errors="coerce"
)

# Выделяем отдельно корректно распознанное время
df_clean["time_parsed"] = (
    df_clean["transaction_datetime"].dt.time
)

# Создаём период в формате «год-месяц»
df_clean["year_month"] = (
    df_clean["date_parsed"].dt.to_period("M")
)

# Определяем последний месяц в датасете
last_date = df_clean["date_parsed"].max()
last_period = last_date.to_period("M")

# Проверяем, заканчивается ли датасет последним днём месяца
last_month_reaches_month_end = last_date.is_month_end

# По умолчанию считаем месяцы полными по границам выгрузки
df_clean["is_full_month_in_dataset"] = True

# Помечаем последний месяц как неполный, если выгрузка закончилась раньше конца месяца
if not last_month_reaches_month_end:
    df_clean.loc[
        df_clean["year_month"] == last_period,
        "is_full_month_in_dataset"
    ] = False

# Получаем час из точного времени операции
df_clean["hour_from_time"] = (
    df_clean["transaction_datetime"].dt.hour
)

# Отмечаем строки, где hour_of_day не совпадает с часом из Time
df_clean["hour_mismatch"] = (
    df_clean["hour_from_time"]
    != df_clean["hour_of_day"]
)

# Считаем количество строк с ошибками преобразования даты или времени
parse_issue_mask = df_clean[
    [
        "date_parsed",
        "transaction_datetime",
        "time_parsed"
    ]
].isna().any(axis=1)

parse_issue_rows = parse_issue_mask.sum()

# Считаем количество удалённых строк
deleted_rows = len(df) - len(df_clean)

# Исходные значения не редактировались, были только добавлены новые столбцы
changed_original_rows = 0

# Проверяем, что исходный df действительно остался без изменений
source_df_unchanged = df.equals(df_original_snapshot)

print("Размер исходного df:", df.shape)
print("Размер подготовленного df_clean:", df_clean.shape)

print("\nИсходный df не изменён:", source_df_unchanged)
print("Удалено строк:", deleted_rows)
print(
    "Строк с изменёнными исходными значениями:",
    changed_original_rows
)
print(
    "Строк с ошибками преобразования даты или времени:",
    parse_issue_rows
)

print(
    "Строк с расхождением hour_of_day и Time:",
    df_clean["hour_mismatch"].sum()
)

print(
    "Строк, относящихся к неполному последнему месяцу:",
    (~df_clean["is_full_month_in_dataset"]).sum()
)

# Повторно проверяем пропуски после подготовки
missing_after_cleaning = pd.DataFrame({
    "missing_count": df_clean.isna().sum(),
    "missing_percent": (
        df_clean.isna().mean() * 100
    ).round(2)
})

print("\nПропуски после подготовки:")
display(missing_after_cleaning)

# Выводим только столбцы, где появились пропуски
columns_with_missing = missing_after_cleaning[
    missing_after_cleaning["missing_count"] > 0
]

print("\nСтолбцы с пропусками после подготовки:")

if columns_with_missing.empty:
    print("Пропуски не обнаружены.")
else:
    display(columns_with_missing)

# Повторно проверяем полные дубликаты по исходным столбцам
duplicates_after_cleaning = df_clean.duplicated(
    subset=df.columns.tolist()
).sum()

print(
    "\nПолные дубликаты после подготовки:",
    duplicates_after_cleaning
)

# Показываем несколько строк с проблемами преобразования, если они есть
if parse_issue_rows > 0:
    print("\nПримеры строк с ошибками преобразования:")
    display(
        df_clean.loc[
            parse_issue_mask,
            [
                "Date",
                "Time",
                "date_parsed",
                "time_parsed",
                "transaction_datetime"
            ]
        ].head()
    )

# Показываем строки с расхождением часа, если они есть
if df_clean["hour_mismatch"].any():
    print("\nСтроки с расхождением часа:")
    display(
        df_clean.loc[
            df_clean["hour_mismatch"],
            [
                "Date",
                "Time",
                "hour_of_day",
                "hour_from_time",
                "coffee_name",
                "money"
            ]
        ]
    )
```

```python
# Создаём период «год-месяц» для корректной месячной динамики
df_clean["year_month"] = (
    df_clean["date_parsed"].dt.to_period("M")
)

# Отмечаем субботу и воскресенье как выходные дни
df_clean["is_weekend"] = (
    df_clean["Weekdaysort"].isin([6, 7])
)

# Находим первую и последнюю даты в датасете
first_date = df_clean["date_parsed"].min()
last_date = df_clean["date_parsed"].max()

# Определяем первый и последний календарные месяцы
first_period = first_date.to_period("M")
last_period = last_date.to_period("M")

# Сначала считаем все месяцы полными
df_clean["is_full_month_in_dataset"] = True

# Помечаем первый месяц неполным, если данные начинаются не 1-го числа
if not first_date.is_month_start:
    df_clean.loc[
        df_clean["year_month"] == first_period,
        "is_full_month_in_dataset"
    ] = False

# Помечаем последний месяц неполным, если данные заканчиваются не последним числом
if not last_date.is_month_end:
    df_clean.loc[
        df_clean["year_month"] == last_period,
        "is_full_month_in_dataset"
    ] = False

# Показываем созданные признаки
display(
    df_clean[
        [
            "Date",
            "date_parsed",
            "year_month",
            "Weekday",
            "Weekdaysort",
            "is_weekend",
            "is_full_month_in_dataset"
        ]
    ].head()
)

# Проверяем типы новых признаков
print(df_clean[
    [
        "year_month",
        "is_weekend",
        "is_full_month_in_dataset"
    ]
].dtypes)

# Проверяем, не появились ли пропуски
print("\nПропуски в новых признаках:")
print(
    df_clean[
        [
            "year_month",
            "is_weekend",
            "is_full_month_in_dataset"
        ]
    ].isna().sum()
)
```

```python
def show_category_summary(data, column, top_n=10):
    # Считаем количество записей по каждой категории
    category_count = data[column].value_counts(dropna=False)

    # Собираем количество и долю категорий в одну таблицу
    category_summary = category_count.rename("count").to_frame()

    category_summary["share_percent"] = (
        category_summary["count"] / len(data) * 100
    ).round(2)

    # Показываем количество уникальных значений
    print(
        f"Количество уникальных значений в {column}:",
        data[column].nunique(dropna=False)
    )

    # Показываем наиболее частые категории
    display(category_summary.head(top_n))
```

```python
# Функция выводит основные характеристики числового столбца
def show_numeric_summary(data, column):
    series = data[column]

    summary = pd.Series({
        "Количество значений": series.count(),
        "Количество пропусков": series.isna().sum(),
        "Среднее": series.mean(),
        "Медиана": series.median(),
        "Минимум": series.min(),
        "Первый квартиль (25%)": series.quantile(0.25),
        "Третий квартиль (75%)": series.quantile(0.75),
        "Максимум": series.max()
    })

    display(summary.to_frame(name=column).round(2))

    # Осторожно оцениваем форму распределения по среднему и медиане
    mean_value = series.mean()
    median_value = series.median()

    if pd.isna(mean_value) or pd.isna(median_value):
        print("Распределение нельзя интерпретировать из-за отсутствующих значений.")
    elif mean_value > median_value:
        print(
            "Среднее выше медианы. В распределении могут присутствовать "
            "более высокие значения, смещающие среднее вверх."
        )
    elif mean_value < median_value:
        print(
            "Среднее ниже медианы. В распределении могут присутствовать "
            "более низкие значения, смещающие среднее вниз."
        )
    else:
        print(
            "Среднее и медиана совпадают. Центральная часть распределения "
            "выглядит относительно сбалансированной."
        )
```

```python
required_columns = ["Time_of_Day", "money"]

if check_columns(df_clean, required_columns):

    # Задаём логический порядок частей дня
    preferred_order = [
        "Morning",
        "Afternoon",
        "Night"
    ]

    # Оставляем только реально существующие категории
    existing_order = [
        category
        for category in preferred_order
        if category in df_clean["Time_of_Day"].dropna().unique()
    ]

    # Добавляем другие категории, если они неожиданно присутствуют
    other_categories = [
        category
        for category in df_clean["Time_of_Day"].dropna().unique()
        if category not in existing_order
    ]

    time_order = existing_order + other_categories

    # Рассчитываем долю операций
    operation_share = (
        df_clean["Time_of_Day"]
        .value_counts(normalize=True)
        .reindex(time_order)
        .mul(100)
    )

    # Рассчитываем долю общей суммы money
    money_share = (
        df_clean
        .groupby("Time_of_Day")["money"]
        .sum()
        .reindex(time_order)
    )

    money_share = money_share / money_share.sum() * 100

    # Создаём позиции для сгруппированных столбцов
    x = np.arange(len(time_order))
    width = 0.35

    # Создаём отдельный график
    plt.figure(figsize=(9, 5))

    bars_operations = plt.bar(
        x - width / 2,
        operation_share.values,
        width,
        label="Доля операций"
    )

    bars_money = plt.bar(
        x + width / 2,
        money_share.values,
        width,
        label="Доля суммы money"
    )

    # Добавляем процентные значения
    plt.bar_label(
        bars_operations,
        fmt="%.1f%%",
        padding=3
    )

    plt.bar_label(
        bars_money,
        fmt="%.1f%%",
        padding=3
    )

    plt.xticks(x, time_order)
    plt.title("Доли операций и суммы money по частям дня")
    plt.xlabel("Часть дня")
    plt.ylabel("Доля, %")
    plt.legend()
    plt.tight_layout()
    plt.show()

    # Определяем лидирующие части дня
    top_operation_period = operation_share.idxmax()
    top_money_period = money_share.idxmax()

    print(
        f"Короткий вывод: наибольшая доля операций приходится "
        f"на {top_operation_period}, а наибольшая доля суммы money "
        f"на {top_money_period}. Категория Night используется "
        "в значении, заданном исходным датасетом."
    )
```

```python
if len(analysis_numeric_columns) >= 2:

    # Оставляем только верхнюю часть матрицы,
    # чтобы каждая пара встречалась один раз
    upper_triangle_mask = np.triu(
        np.ones(
            correlation_matrix.shape,
            dtype=bool
        ),
        k=1
    )

    # Преобразуем матрицу в таблицу пар признаков
    correlation_pairs = (
        correlation_matrix
        .where(upper_triangle_mask)
        .stack()
        .reset_index()
    )

    correlation_pairs.columns = [
        "feature_1",
        "feature_2",
        "correlation"
    ]

    # Рассчитываем абсолютную силу связи
    correlation_pairs["absolute_correlation"] = (
        correlation_pairs["correlation"].abs()
    )

    # Считаем число наблюдений для каждой пары
    correlation_pairs["observation_count"] = (
        correlation_pairs.apply(
            lambda row: df_clean[
                [row["feature_1"], row["feature_2"]]
            ].dropna().shape[0],
            axis=1
        )
    )

    # Сортируем пары от наиболее сильной связи
    correlation_pairs = (
        correlation_pairs
        .sort_values(
            "absolute_correlation",
            ascending=False
        )
        .reset_index(drop=True)
    )

    print("\nПары с наиболее сильной корреляцией:")
    display(
        correlation_pairs[
            [
                "feature_1",
                "feature_2",
                "correlation",
                "absolute_correlation",
                "observation_count"
            ]
        ].head(10)
    )
```

```python
import pandas as pd
import numpy as np

# Проверяем наличие необходимых столбцов
required_columns = ["coffee_name", "money"]

missing_columns = [
    column
    for column in required_columns
    if column not in df_clean.columns
]

if missing_columns:
    print(
        "ABC-анализ не выполнен. Отсутствуют столбцы:",
        missing_columns
    )

else:
    # Группируем данные по напиткам и считаем суммарный вклад
    abc_table = (
        df_clean
        .groupby("coffee_name", as_index=False)
        .agg(
            total_money=("money", "sum"),
            transaction_count=("money", "size")
        )
    )

    # Сортируем напитки по убыванию суммарного вклада
    abc_table = (
        abc_table
        .sort_values(
            by="total_money",
            ascending=False
        )
        .reset_index(drop=True)
    )

    # Рассчитываем общую сумму money
    total_money_all = abc_table["total_money"].sum()

    # Проверяем, что общий вклад положительный
    if total_money_all <= 0:
        print(
            "ABC-анализ не выполнен: "
            "общая сумма money должна быть больше нуля."
        )

    else:
        # Рассчитываем долю каждого напитка в общей сумме
        abc_table["share"] = (
            abc_table["total_money"]
            / total_money_all
        )

        # Рассчитываем накопленную долю
        abc_table["cumulative_share"] = (
            abc_table["share"].cumsum()
        )

        # Присваиваем классы по накопленной доле
        abc_table["ABC_class"] = np.select(
            [
                abc_table["cumulative_share"] <= 0.80,
                abc_table["cumulative_share"] <= 0.95
            ],
            [
                "A",
                "B"
            ],
            default="C"
        )

        # Создаём процентные столбцы для удобного отображения
        abc_table["share_percent"] = (
            abc_table["share"] * 100
        ).round(2)

        abc_table["cumulative_share_percent"] = (
            abc_table["cumulative_share"] * 100
        ).round(2)

        # Округляем суммарный денежный вклад
        abc_table["total_money"] = (
            abc_table["total_money"].round(2)
        )

        # Выводим итоговую таблицу
        display(
            abc_table[
                [
                    "coffee_name",
                    "transaction_count",
                    "total_money",
                    "share_percent",
                    "cumulative_share_percent",
                    "ABC_class"
                ]
            ]
        )
```

```python
# SAVE_PROJECT_TABLES
from pathlib import Path
import pandas as pd

current_dir = Path.cwd()
project_root = (
    current_dir.parent
    if current_dir.name == "notebooks"
    else current_dir
)

outputs_dir = project_root / "outputs"
charts_dir = project_root / "charts"
data_clean_dir = project_root / "data_clean"

outputs_dir.mkdir(parents=True, exist_ok=True)
charts_dir.mkdir(parents=True, exist_ok=True)
data_clean_dir.mkdir(parents=True, exist_ok=True)


def prepare_table_for_csv(table, preserve_index=False):
    export_table = table.copy()

    if preserve_index:
        export_table = export_table.reset_index()
        if "index" in export_table.columns:
            export_table = export_table.rename(
                columns={"index": "feature"}
            )

    for column in export_table.columns:
        dtype = export_table[column].dtype

        if isinstance(dtype, pd.PeriodDtype):
            export_table[column] = export_table[column].astype(str)

        if isinstance(dtype, pd.CategoricalDtype):
            export_table[column] = export_table[column].astype(str)

    rename_map = {
        "transaction_count": "observation_count",
        "transaction_share_percent": "observation_share_percent",
        "daily_transaction_count": "daily_observation_count",
        "mean_daily_transactions": "mean_daily_observations",
        "median_daily_transactions": "median_daily_observations",
        "min_daily_transactions": "min_daily_observations",
        "max_daily_transactions": "max_daily_observations",
        "total_transactions": "total_observations",
    }

    return export_table.rename(
        columns={
            old_name: new_name
            for old_name, new_name in rename_map.items()
            if old_name in export_table.columns
        }
    )


# Очищенный датасет
df_export = df_clean.copy()

for column in df_export.columns:
    dtype = df_export[column].dtype

    if isinstance(dtype, pd.PeriodDtype):
        df_export[column] = df_export[column].astype(str)

    if isinstance(dtype, pd.CategoricalDtype):
        df_export[column] = df_export[column].astype(str)

if "time_parsed" in df_export.columns:
    df_export["time_parsed"] = df_export["time_parsed"].astype(str)

clean_path = data_clean_dir / "coffee_sales_cleaned.csv"

df_export.to_csv(
    clean_path,
    index=False,
    encoding="utf-8-sig",
    date_format="%Y-%m-%d %H:%M:%S.%f"
)


table_exports = [
    ("coffee_summary", "01_coffee_summary.csv", False),
    ("hour_summary", "02_hour_summary.csv", False),
    ("time_of_day_summary", "03_time_of_day_summary.csv", False),
    ("weekday_summary", "04_weekday_summary.csv", False),
    ("monthly_summary", "05_monthly_summary.csv", False),
    ("coffee_money_summary", "06_coffee_money_ranking.csv", False),
    ("correlation_matrix", "07_correlation_matrix.csv", True),
    ("correlation_pairs", "08_correlation_pairs.csv", False),
    ("abc_table", "09_coffee_abc_analysis.csv", False),
    ("abc_class_summary", "10_abc_class_summary.csv", False),
]

saved_tables = []

for variable_name, filename, preserve_index in table_exports:
    if variable_name not in globals():
        continue

    export_table = prepare_table_for_csv(
        globals()[variable_name],
        preserve_index=preserve_index
    )

    export_table.to_csv(
        outputs_dir / filename,
        index=False,
        encoding="utf-8-sig"
    )

    saved_tables.append(filename)

print("Сохранён очищенный CSV:")
print("-", clean_path.name)

print("\nСохранены итоговые таблицы:")
for filename in saved_tables:
    print("-", filename)
```
